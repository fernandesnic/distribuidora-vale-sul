## Conferências entre tabelas

Feitas depois de explorar todas as tabelas. Primeiro a integridade referencial (todo `_id` precisa existir na tabela de origem), depois as regras de negócio.

### Integridade referencial

```python
def orfaos(tabela, coluna, origem, coluna_origem):
    """Conta os valores da coluna que não existem na tabela de origem (ignora nulos)."""
    valores = tabela[coluna].dropna()
    return (~valores.isin(origem[coluna_origem])).sum()

checagens = [
    ("clientes.vendedor_id", clientes, "vendedor_id", vendedores, "vendedor_id"),
    ("pedidos.cliente_id", pedidos, "cliente_id", clientes, "cliente_id"),
    ("pedidos.vendedor_id", pedidos, "vendedor_id", vendedores, "vendedor_id"),
    ("itens.pedido_id", itens, "pedido_id", pedidos, "pedido_id"),
    ("itens.produto_id", itens, "produto_id", produtos, "produto_id"),
    ("titulos.pedido_id", titulos, "pedido_id", pedidos, "pedido_id"),
    ("titulos.cliente_id", titulos, "cliente_id", clientes, "cliente_id"),
    ("pagamentos.titulo_id", pagamentos, "titulo_id", titulos, "titulo_id"),
]
for nome, t, c, o, co in checagens:
    print(f"{nome:25} órfãos: {orfaos(t, c, o, co)}")
```

```python
itens[~itens["produto_id"].isin(produtos["produto_id"])]
```

```python
# Juros só quando o pagamento atrasa
pag_tit = pagamentos.assign(data_pag=data_pagamento).merge(titulos[["titulo_id", "data_vencimento"]], on="titulo_id")
atrasado = pag_tit["data_pag"] > pd.to_datetime(pag_tit["data_vencimento"])
print("juros sem atraso:", ((pag_tit["valor_juros"] > 0) & ~atrasado).sum())
sem_juros = pag_tit[atrasado & (pag_tit["valor_juros"] == 0) & (pag_tit["forma_pagamento"] == "BOLETO")]
dias_atraso = (sem_juros["data_pag"] - pd.to_datetime(sem_juros["data_vencimento"])).dt.days
venc_domingo = pd.to_datetime(sem_juros["data_vencimento"]).dt.day_of_week == 6
print("boleto atrasado sem juros:", len(sem_juros))
print("  1 dia de atraso com vencimento no domingo:", ((dias_atraso == 1) & venc_domingo).sum())
print("  os demais:", (~((dias_atraso == 1) & venc_domingo)).sum())
n_pag = pagamentos[pagamentos["forma_pagamento"] == "BOLETO"].drop_duplicates().groupby("titulo_id").size()
outros = sem_juros[~((dias_atraso == 1) & venc_domingo)]
print("  dos demais, em títulos com status PARCIAL ou pagos em mais de um boleto:",
      (outros["titulo_id"].map(n_pag).gt(1) | outros["titulo_id"].isin(titulos.loc[titulos["status"] == "PARCIAL", "titulo_id"])).sum())
```

  - **361** são de títulos pagos em partes (status `PARCIAL` ou mais de um boleto): as parcelas pagas em atraso foram registradas sem juros
- **Ponto de atenção:** para medir atraso, considerar a prorrogação do fim de semana. Os juros das parcelas não são confiáveis para medir quanto se ganhou com atraso
