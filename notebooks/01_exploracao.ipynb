três pedidos, 133467, 143683, 156644 possuem data com erro de digitação, corrigiria colocando em 2025.

```python
data_faturamento = pd.to_datetime(pedidos["data_faturamento"])
data_faturamento.describe()
```

- **Decisão:** o período das vendas é definido pela **data de faturamento**, que é quando a venda vira nota e conta no financeiro. Assim os números batem com os relatórios financeiros
  - Consequência: pedidos cancelados (sem faturamento) ficam fora das vendas automaticamente
- Faturamento máximo em 02/01/2026: normal, é pedido do fim de dezembro faturado na virada do ano

```python
pedidos[data_faturamento < data_pedido]
```

- Nenhum pedido foi faturado antes de ser feito, fora os 3 com ano 2052 (erro de digitação na `data_pedido`)

```python
diferenca = (pedidos["valor_produtos"] - pedidos["valor_desconto"]) - pedidos["valor_total"]
pedidos[diferenca.abs() > 0.01]
```

- A conta `valor_produtos - valor_desconto = valor_total` bate em todos os pedidos (tolerância de 1 centavo)
